# NaN Analysis

This notebook analyzes the remaining NaN values in the filled datasets from `data/processed/dataset_features_removed` directory to ensure data quality.

## Setup and Data Loading

We'll use numpy for data analysis and matplotlib for visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Define the absolute data directory
BASE_DIR = '/Users/georgialex/Library/Mobile Documents/com~apple~CloudDocs/EPFL/Master_Data_Science/ML/Project1/project-1-heartbreakers'
data_dir = os.path.join(BASE_DIR, 'data', 'processed', 'dataset_features_removed')

# Load the datasets using numpy
train_data = np.genfromtxt(
    os.path.join(data_dir, 'x_train_filled.csv'),
    delimiter=',',
    skip_header=1  # Skip header row
)

test_data = np.genfromtxt(
    os.path.join(data_dir, 'x_test_filled.csv'),
    delimiter=',',
    skip_header=1  # Skip header row
)

# Get column names from the files
with open(os.path.join(data_dir, 'x_train_filled.csv'), 'r') as f:
    header = f.readline().strip().split(',')

# Function to analyze NaN values
def analyze_nan_values(data, dataset_name):
    # Count NaN values per column
    nan_counts = np.isnan(data).sum(axis=0)
    total_samples = data.shape[0]
    
    # Find columns with NaN values
    cols_with_nans = np.where(nan_counts > 0)[0]
    
    if len(cols_with_nans) == 0:
        print(f"\nNo NaN values found in {dataset_name}!")
        return
    
    print(f"\nNaN values found in {dataset_name}:")
    print("=" * 50)
    
    # Print details for each column with NaNs
    for col_idx in cols_with_nans:
        count = nan_counts[col_idx]
        percentage = (count / total_samples) * 100
        feature_name = header[col_idx] if col_idx < len(header) else f"Column {col_idx}"
        print(f"{feature_name}: {count} NaN values ({percentage:.2f}%)")
    
    # Print summary
    total_nans = np.isnan(data).sum()
    total_elements = data.size
    overall_percentage = (total_nans / total_elements) * 100
    print(f"\nOverall summary for {dataset_name}:")
    print(f"Total NaN values: {total_nans}")
    print(f"Total elements: {total_elements}")
    print(f"Overall NaN percentage: {overall_percentage:.2f}%")

# Analyze training data
print("Analyzing training data...")
analyze_nan_values(train_data, "Training Set")

print("\n" + "=" * 50 + "\n")

# Analyze test data
print("Analyzing test data...")
analyze_nan_values(test_data, "Test Set")
```
